```python
import pandas as pd
df = pd.read_csv(r"Online_Retail_ucl.csv",  encoding='ISO-8859-1')
df.head()
```

```python
df.dropna(subset=["CustomerID"], inplace=True)   #dropping rows with customer data
df = df[df["Quantity"] > 0]  #keeping only positive sales
df = df[df["UnitPrice"] > 0]  ##removing refunds and returns
```

```python
df.info()
```

```python
df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"], format="%d-%m-%Y %H:%M")
```

```python
#Mark items as discounted if their price is below the typical price for the same product
df["PromoFlag"] = df["UnitPrice"] < df.groupby("StockCode")["UnitPrice"].transform("median")
```

```python
df.head()
```

```python
df.tail()
```

```python
import statsmodels.formula.api as smf
```

The following code is used to understand the factors that affect the quantity sold, such as pricing (UnitPrice), promotions (PromoFlag), product types (StockCode), and sales locations (Country).

```python
#Sample a subset to speed up modeling
sample_df = df.sample(50000, random_state=42)

#Building regression model
model = smf.ols("Quantity ~ UnitPrice + PromoFlag + C(Country) + C(StockCode)", data=sample_df).fit()
print(model.summary())
```

***Visualize Insights***

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Filter and process the promo data
promo_data = df[df["StockCode"].isin(df["StockCode"].value_counts().head(20).index)].copy()
promo_data["PromoFlag"] = promo_data["PromoFlag"].astype(int)

# Set a larger plot size
plt.figure(figsize=(12, 8))  # Adjust width and height as needed

# Create the boxplot
sns.boxplot(x="PromoFlag", y="Quantity", data=promo_data)

# Customize the title and x-axis labels
plt.title("Units Sold with and without Promo", fontsize=18)
plt.xticks([0, 1], ["No Promo", "Promo"], fontsize=12)
plt.ylabel("Quantity", fontsize=14)
plt.xlabel("Promo Flag", fontsize=14)

# Display the plot
plt.show()
```

```python
sns.boxplot(
    y="Quantity",
    x="PromoFlag",
    hue="PromoFlag",
    data=promo_data,
    palette={0: "gray", 1: "skyblue"},
    showfliers=True,
    legend=False
)
```

```python
import numpy as np
import statsmodels.formula.api as smf

# You can also log-transform Quantity if needed for better normality
promo_data["Log_Quantity"] = np.log1p(promo_data["Quantity"])

# Regression model
model = smf.ols("Log_Quantity ~ UnitPrice + PromoFlag", data=promo_data).fit()
print(model.summary())
```

```python
plt.figure(figsize=(10, 6))
sns.boxplot(x="PromoFlag", y="Log_Quantity", data=promo_data, hue="PromoFlag", palette="Set2", legend=False)
plt.title("Log_Quantity Sold: Promo vs. Non-Promo", fontsize=16)
plt.xticks([0, 1], ["No Promo", "Promo"], fontsize=12)
plt.ylabel("Log_Quantity", fontsize=14)
plt.xlabel("Promotion Status", fontsize=14)
plt.show()
```

- Key Findings:- Promotions have a very strong positive effect on sales quantity.
- Price has a slight negative impact, consistent with elasticity theory (higher prices reduce demand).
